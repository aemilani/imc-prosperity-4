```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# ROUND 1

```python
with open(f'../leaderboard/raw/ALGO_2026-04-19_15-42-08.json', 'r') as f:
    algo_data_1 = json.load(f)
    algo_data_1 = pd.json_normalize(algo_data_1)
    algo_data_1 = algo_data_1.set_index('team.id', drop=True)
with open(f'../leaderboard/raw/MANUAL_2026-04-19_15-45-55.json', 'r') as f:
    manual_data_1 = json.load(f)
    manual_data_1 = pd.json_normalize(manual_data_1)
    manual_data_1 = manual_data_1.set_index('team.id', drop=True)
with open(f'../leaderboard/raw/OVERALL_2026-04-19_15-38-20.json', 'r') as f:
    overall_data_1 = json.load(f)
    overall_data_1 = pd.json_normalize(overall_data_1)
    overall_data_1 = overall_data_1.set_index('team.id', drop=True)
```

```python
print(f"Number of participants: {len(overall_data_1)}")
```

```python
algo_pnls = algo_data_1['score'].to_numpy()
my_algo_pnl = algo_data_1[algo_data_1['team.name'] == 'Arbland']['score'].to_numpy()[0]

manual_pnls = manual_data_1['score'].to_numpy()
my_manual_pnl = manual_data_1[manual_data_1['team.name'] == 'Arbland']['score'].to_numpy()[0]

overall_pnls = overall_data_1['score'].to_numpy()
my_overall_pnl = overall_data_1[overall_data_1['team.name'] == 'Arbland']['score'].to_numpy()[0]
```

```python
print(f"My Algo position: Top {algo_data_1[algo_data_1['team.name'] == 'Arbland']['position'].to_numpy()[0] / len(algo_data_1) * 100:.2f}%")
print(f"My Manual position: Top {manual_data_1[manual_data_1['team.name'] == 'Arbland']['position'].to_numpy()[0] / len(manual_data_1) * 100:.2f}%")
print(f"My Overall position: Top {overall_data_1[overall_data_1['team.name'] == 'Arbland']['position'].to_numpy()[0] / len(overall_data_1) * 100:.2f}%")
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 3))
ax[0].hist([pnl for pnl in algo_pnls if pnl > 0], bins=70)
ax[0].axvline(my_algo_pnl, c='r')
ax[0].set_xlabel('PnL')
ax[0].set_title('Profitable Algo PnLs')
ax[1].hist([pnl for pnl in manual_pnls if pnl > 0], bins=70)
ax[1].axvline(my_manual_pnl, c='r')
ax[1].set_xlabel('PnL')
ax[1].set_title('Profitable Manual PnLs')
ax[2].hist([pnl for pnl in overall_pnls if pnl > 0], bins=70)
ax[2].axvline(my_overall_pnl, c='r')
ax[2].set_xlabel('PnL')
ax[2].set_title('Profitable Overall PnLs')
plt.show()
```

```python
print(f"Median ALgo PnL: {np.median(algo_pnls)}")
print(f"Median Non-zero ALgo PnL: {np.median([pnl for pnl in algo_pnls if pnl != 0])}")
```

```python
print(f"Ratio of profitable algos: {len([pnl for pnl in algo_pnls if pnl > 0]) / len(algo_pnls)}")
```

# Selected teams

```python
print(f"Dropped out team: {overall_data_1.loc[list(set(overall_data_1.index) - set(overall_data_2.index))[0]]['team.name']}")

new_ids = list(set(overall_data_2.index) - set(overall_data_1.index))
print(f"Number of teams joining in round 2: {len(new_ids)}")
```

```python
algo_data_1 = pd.concat([algo_data_1, pd.DataFrame(index=new_ids)])
manual_data_1 = pd.concat([manual_data_1, pd.DataFrame(index=new_ids)])
overall_data_1 = pd.concat([overall_data_1, pd.DataFrame(index=new_ids)])
```

```python
selected_teams = pd.DataFrame()
selected_teams.loc[:, 'position'] = overall_data_2[overall_data_2['score'] >= 200000]['position']
selected_teams.loc[:, 'team.name'] = overall_data_2[overall_data_2['score'] >= 200000]['team.name']
selected_teams.loc[:, 'team.country'] = overall_data_2[overall_data_2['score'] >= 200000]['team.countryCode']

selected_teams.loc[:, 'r1_algo_score'] = algo_data_1.loc[selected_teams.index, 'score']
selected_teams.loc[:, 'r1_manual_score'] = manual_data_1.loc[selected_teams.index, 'score']
selected_teams.loc[:, 'r1_overall_score'] = overall_data_1.loc[selected_teams.index, 'score']
selected_teams.loc[:, 'r2_algo_score'] = algo_data_2.loc[selected_teams.index, 'score']
selected_teams.loc[:, 'r2_manual_score'] = manual_data_2.loc[selected_teams.index, 'score']
selected_teams.loc[:, 'r2_overall_score'] = overall_data_2.loc[selected_teams.index, 'score']
```

```python
algo_pnls_selected = selected_teams['r2_algo_score'].to_numpy()
manual_pnls_selected = selected_teams['r2_manual_score'].to_numpy()
overall_pnls_selected = selected_teams['r2_overall_score'].to_numpy()
```

```python
print(f"My Algo position: Top {algo_data_2[algo_data_2['team.name'] == 'Arbland']['position'].to_numpy()[0] / len(selected_teams) * 100:.2f}%")
print(f"My Manual position: Top {manual_data_2[manual_data_2['team.name'] == 'Arbland']['position'].to_numpy()[0] / len(selected_teams) * 100:.2f}%")
print(f"My Overall position: Top {overall_data_2[overall_data_2['team.name'] == 'Arbland']['position'].to_numpy()[0] / len(selected_teams) * 100:.2f}%")
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 3))
ax[0].hist([pnl for pnl in algo_pnls_selected if pnl > 0], bins=70)
ax[0].axvline(my_algo_pnl, c='r')
ax[0].set_xlabel('PnL')
ax[0].set_title('Selected Algo PnLs')
ax[1].hist([pnl for pnl in manual_pnls_selected if pnl > 0], bins=70)
ax[1].axvline(my_manual_pnl, c='r')
ax[1].set_xlabel('PnL')
ax[1].set_title('Selected Manual PnLs')
ax[2].hist([pnl for pnl in overall_pnls_selected if pnl > 0], bins=70)
ax[2].axvline(my_overall_pnl, c='r')
ax[2].set_xlabel('PnL')
ax[2].set_title('Selected Overall PnLs')
plt.show()
```

```python
# Selected teams' round 2 manual challenge PnL
manual_pnls_round_2_only = selected_teams['r2_manual_score'] - selected_teams['r1_manual_score']
my_manual_pnl_round_2_only = manual_pnls_round_2_only.loc[overall_data_2[overall_data_2['team.name'] == 'Arbland'].index].to_numpy()[0]
```
