```python
import json
import numpy as np
import matplotlib.pyplot as plt
```

# ROUND 1

```python
with open(f'../leaderboard/raw/ALGO_2026-04-19_15-42-08.json', 'r') as f:
    algo_data = json.load(f)
with open(f'../leaderboard/raw/MANUAL_2026-04-19_15-45-55.json', 'r') as f:
    manual_data = json.load(f)
with open(f'../leaderboard/raw/OVERALL_2026-04-19_15-38-20.json', 'r') as f:
    overall_data = json.load(f)
```

```python
algo_pnls = [d['score'] for d in algo_data]
my_algo_pnl = [d['score'] for d in algo_data if d['team']['name'] == 'Arbland'][0]

manual_pnls = [d['score'] for d in manual_data]
my_manual_pnl = [d['score'] for d in manual_data if d['team']['name'] == 'Arbland'][0]

overall_pnls = [d['score'] for d in overall_data]
my_overall_pnl = [d['score'] for d in overall_data if d['team']['name'] == 'Arbland'][0]
```

```python
print(f"My Algo position: Top {[d['position'] for d in algo_data if d['team']['name'] == 'Arbland'][0] / len(algo_data) * 100:.2f}%")
print(f"My Manual position: Top {[d['position'] for d in manual_data if d['team']['name'] == 'Arbland'][0] / len(manual_data) * 100:.2f}%")
print(f"My Overall position: Top {[d['position'] for d in overall_data if d['team']['name'] == 'Arbland'][0] / len(overall_data) * 100:.2f}%")
```

```python
fig, ax = plt.subplots(1, 3, figsize=(15, 3))
ax[0].hist([pnl for pnl in algo_pnls if pnl > 0], bins=70)
ax[0].axvline(my_algo_pnl, c='r')
ax[0].set_xlabel('PnL')
ax[0].set_title('Profitable Algo PnLs')
ax[1].hist([pnl for pnl in manual_pnls if pnl > 0], bins=70)
ax[1].axvline(my_manual_pnl, c='r')
ax[1].set_xlabel('PnL')
ax[1].set_title('Profitable Manual PnLs')
ax[2].hist([pnl for pnl in overall_pnls if pnl > 0], bins=70)
ax[2].axvline(my_overall_pnl, c='r')
ax[2].set_xlabel('PnL')
ax[2].set_title('Profitable Overall PnLs')
plt.show()
```

```python
print(f"Median ALgo PnL: {np.median(algo_pnls)}")
print(f"Median Non-zero ALgo PnL: {np.median([pnl for pnl in algo_pnls if pnl != 0])}")
```
